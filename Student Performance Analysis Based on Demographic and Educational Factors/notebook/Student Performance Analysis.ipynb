## 6. Create Visualizations


```python
#histogram of math scores
plt.hist(df['math score'], bins=10, color='blue', alpha=0.7)
plt.title('Distribution of Math Scores')
plt.xlabel('Math Score')
plt.ylabel('Number of Students')
plt.show()
```

Distribution of Math Scores

```python
#Average Math Score by Gender
df.groupby("gender")["math score"].mean().plot(kind="bar")
plt.title("Average Math Score by Gender")
plt.show()
```

Compare average scores by gender.

```python
#Average Reading Score by Test Preparation
df.groupby("test preparation course")["reading score"].mean().plot(kind="bar")
plt.show()
```

Compare students who completed the course with those who did not.

```python
#histogram of reading scores
plt.hist(df['reading score'], bins=10, color='green', alpha=0.7)
plt.title('Distribution of Reading Scores')
plt.xlabel('Reading Score')
plt.ylabel('Number of Students')
plt.show()
```

Distribution of Reading scores

```python
#Average Reading Score by Gender
df.groupby("gender")["reading score"].mean().plot(kind="bar")
plt.title("Average Reading Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Reading Score")
plt.show()
```

Compare average Reading scores by gender.

```python
#histogram of writing scores
plt.hist(df['writing score'], bins=10, color='green', alpha=0.7)
plt.title('Distribution of Writing Scores')
plt.xlabel('Writing Score')
plt.ylabel('Number of Students')
plt.show()
```

Distribution of writing scores

```python
#Average Writing Score by Gender
df.groupby("gender")["writing score"].mean().plot(kind="bar")
plt.title("Average Writing Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Writing Score")
plt.show()
```

Compare average writong scores by gender.

```python
df[["math score","reading score","writing score"]].corr()
```

#### The analysis produced several important findings.

- The dataset contains 1,000 student records with no missing values.
- Reading and writing scores have the strongest positive correlation (r = 0.949), indicating that students who perform well in reading also tend to perform well in writing.
- Mathematics scores also show strong positive correlations with reading (r = 0.812) and writing (r = 0.790) scores.
- Students who completed the test preparation course generally achieved higher exam scores than those who did not.
- Students whose parents had higher education levels tended to achieve better academic performance.
- Student performance varied across demographic and educational factors.

## 2: The API Challenge


### The "Data Collection Problem"


During the planning stage, we explored using APIs to collect educational data. However, several challenges made this approach less practical.

- Limited Student Data: Most education APIs provide national or regional statistics rather than individual student records.
- API Restrictions: Some APIs require authentication, API keys, or have request limits.
- Complex Data Processing: API responses are usually in JSON format and require additional processing before analysis.
- Time Constraints: Collecting and cleaning API data would require significantly more time than using a prepared dataset.

## 3: The Pivot to Kaggle Dataset


### Why We Changed Our Approach


To focus on data analysis instead of data collection, we decided to use a pre-collected dataset from Kaggle.

#### **Dataset Used:**

`exams.xlsx`

#### **Benefits**

- **Ready to Use:** 1,000 student records available immediately.
- **Complete Data:** No missing values, making preprocessing easier.
- **Reliable:** Suitable for statistical analysis and visualization.
- **Rich Features:** Includes demographic information, parental education, lunch type, test preparation course, and exam scores.
